```python
import matplotlib.pyplot as plt

# 1. Tes données
langages = [ "Python", "SQL", "R", "Java"]
popularité = [85, 70, 40, 60]

# 2. Création du graphique
plt.bar(langages, popularité, color="skyblue")
plt.title("Langages les plus utilisés en Data Analysis")
plt.xlabel("Langages")
plt.ylabel("Score d'utilisation")

# 3. Affichage
plt.show()
```

```python
import pandas as pd
# Création d'un mini jeu de données (DataFrame)
data = {
    "Projet" : ["Data Analysis", "IA Générative", "Machine Learning", "Automation", "Dashboard"],
    "Niveau" : ["Débutant", "Avancé", "Intermédiaire", "Débutant", "Intermédiaire"],
    "Heure_Etude": [15, 40, 30, 10, 20],
    "Complété" : [True, False, False, True, True]
    }

# 2. Conversion en tableau Pandas
df = pd.DataFrame(data)

# 3. Afficher le tableau complet
print("--- Mon premier Tableau Pandas ---")
display(df)

# 4. Faire de petites analyses
print("\n--- Statistique sur les heures d'étude ---")
print("Temps moyen d'étude :", df["Heure_Etude"].mean(), "heures")
print("Temps totale d'étude :", df["Heure_Etude"].sum(), "heures")

# 5. Filtrer les données (ex : voir seulement les projets "Débutant")
print("\n--- Projets niveau Débutant ---")
display(df[df["Niveau"] == "Débutant"])
```

```python
import matplotlib.pyplot as plt

# Graphique à barre à partir du tableau Pandas
plt.figure(figsize=(8, 4))
plt.bar(df["Projet"], df['Heure_Etude'], color="#4C72B0")

plt.title("Heures consacrées par projet")
plt.xlabel("Projets")
plt.ylabel("Heures d'étude")
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.show()
```
